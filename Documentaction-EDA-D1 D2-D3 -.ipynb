# Databricks Lakehouse EDA – Sales & Customer Insights

An End-to-End Exploratory Data Analysis (EDA) project built natively on the Databricks Lakehouse platform using SQL and Delta Lake architecture.

## 📌 Project Overview
This project transforms raw transactional data into structured, clean analytics-ready tables to evaluate company health, product category behaviors, and customer demographics. The entire process follows data engineering best practices, establishing a dedicated catalog, data isolation layers, syntax troubleshooting, and an executive-ready dashboard.

---

## 🏗️ Data Architecture (Medallion Structure)
Instead of utilizing shared course environments, a completely isolated catalog named `sales_project_db` was deployed to manage data lifecycle over two explicit layers:

1. **Bronze Layer (`sales_project_db.bronze`)**: Raw, ingestion-level table replicas directly mirroring historical event data.
2. **Silver Layer (`sales_project_db.silver`)**: Fully cleaned, decoupled dimensional schemas containing standardized data types, concatenated attributes (e.g., customer full names), and verified foreign keys ready for consumption.

* **Double-Dot Path Failures:** Caught and eliminated standard compiler syntax bugs (`[PARSE_SYNTAX_ERROR]`) generated by unintended dual delimiters (`catalog..table`).
