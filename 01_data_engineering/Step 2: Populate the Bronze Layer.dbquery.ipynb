```sql
-- Ingesting sales fact data into Bronze
CREATE TABLE IF NOT EXISTS sales_project_db.bronze.raw_sales
AS SELECT * FROM workspace.salesdb.fact_sales;

-- Ingesting products dimensional data into Bronze
CREATE TABLE IF NOT EXISTS sales_project_db.bronze.raw_products
AS SELECT * FROM workspace.salesdb.dim_products;

-- Ingesting customers dimensional data into Bronze
CREATE TABLE IF NOT EXISTS sales_project_db.bronze.raw_customers
AS SELECT * FROM workspace.salesdb.dim_customers;


```
