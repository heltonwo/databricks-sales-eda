```sql
-- 1. Silver Products (Cleaned dimensional data)
CREATE TABLE IF NOT EXISTS sales_project_db.silver.dim_products
AS SELECT 
    product_key,
    product_name,
    category
FROM sales_project_db.bronze.raw_products;

-- 2. Silver Customers (Data cleansing and name concatenation)
CREATE TABLE IF NOT EXISTS sales_project_db.silver.dim_customers
AS SELECT 
    customer_key,
    customer_number,
    CONCAT(first_name, ' ', last_name) AS customer_name,
    birthdate
FROM sales_project_db.bronze.raw_customers;

-- 3. Silver Sales (Clean fact table mapped to your catalog)
CREATE TABLE IF NOT EXISTS sales_project_db.silver.fact_sales
AS SELECT 
    order_date,
    order_number,
    product_key,
    customer_key,
    sales_amount,
    quantity,
    price
FROM sales_project_db.bronze.raw_sales;
```
