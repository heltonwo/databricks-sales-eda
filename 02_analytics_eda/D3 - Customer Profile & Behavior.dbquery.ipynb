```sql
SELECT
  CASE 
    WHEN customer_age < 30 THEN 'Under 30'
    WHEN customer_age BETWEEN 30 AND 39 THEN '30-39'
    WHEN customer_age BETWEEN 40 AND 49 THEN '40-49'
    WHEN customer_age BETWEEN 50 AND 59 THEN '50-59'
    ELSE '60 and over'
  END AS age_group,
  COUNT(DISTINCT customer_number) AS total_customers,
  ROUND(SUM(total_customer_spend), 2) AS total_revenue,
  ROUND(AVG(total_customer_spend), 2) AS average_spend_per_customer
FROM (
  SELECT
    c.customer_number,
    YEAR(CURRENT_DATE()) - YEAR(c.birthdate) AS customer_age,
    SUM(f.sales_amount) AS total_customer_spend
  FROM sales_project_db.silver.fact_sales f
  INNER JOIN sales_project_db.silver.dim_customers c
    ON f.customer_key = c.customer_key
  WHERE c.birthdate IS NOT NULL
  GROUP BY 1, 2
)
GROUP BY 1
ORDER BY age_group ASC;
```
