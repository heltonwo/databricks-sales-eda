```python
# D1 - Sales Performance & Revenue
# Source: sales_project_db.silver.fact_sales
# PySpark equivalent of the original SQL query (D1 - Sales Performance & Revenue.dbquery)
 
from pyspark.sql import functions as f
 
fact_sales = spark.table("sales_project_db.silver.fact_sales")
 
df_sales_performance = (
    fact_sales
    .filter(f.col("order_date").isNotNull())
    .withColumn("order_month", f.date_trunc("month", f.col("order_date")))
    .groupBy("order_month")
    .agg(
        f.countDistinct("order_number").alias("total_orders"),
        f.round(f.sum("sales_amount"), 2).alias("total_revenue"),
        f.sum("quantity").alias("total_items_sold"),
        f.round(
            f.sum("sales_amount") / f.countDistinct("order_number"), 2
        ).alias("average_ticket"),
    )
    .orderBy(f.col("order_month").asc())
)
 
display(df_sales_performance)
```
