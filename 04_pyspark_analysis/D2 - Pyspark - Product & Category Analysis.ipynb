```python
# D2 - Product & Category Analysis
# Sources: sales_project_db.silver.fact_sales, sales_project_db.silver.dim_products
# PySpark equivalent of the original SQL query (D2 - Product & Category Analysis.dbquery)
 
from pyspark.sql import functions as f
 
fact_sales = spark.table("sales_project_db.silver.fact_sales")
dim_products = spark.table("sales_project_db.silver.dim_products")
 
df_product_category = (
    fact_sales.alias("f")
    .join(
        dim_products.alias("p"),
        f.col("f.product_key") == f.col("p.product_key"),
        "inner",
    )
    .groupBy(
        f.col("p.category").alias("product_category"),
        f.col("p.product_name").alias("product_name"),
    )
    .agg(
        f.count("*").alias("total_sales_records"),
        f.sum("f.quantity").alias("units_sold"),
        f.round(f.sum("f.sales_amount"), 2).alias("total_revenue"),
    )
    .orderBy(f.col("total_revenue").desc())
)
 
display(df_product_category)
```
